```python
# 导包
import pandas as pd
import seaborn as sns  # 底层依赖 matplotlib
import matplotlib.pyplot as plt
from IPython.core.pylabtools import figsize

# 展示中文
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS']
plt.rcParams['axes.unicode_minus'] = False     # 正常显示负号
```

## 1. 扩展_Pandas绘图

```python
# 介绍：Pandas绘图可以基于 df对象直接绘图，它的底层依赖 Matplotlib
```

```python
# 1. 加载数据集，并设置第1列为 索引列
df = pd.read_csv('./data/winemag-data_first150k.csv', index_col=0)
df
```

```python
# 2. 查看数据集
# winemag-data_first150k.csv9数据集的字段介绍如下：
# 1.country：国家
# 2.description：描述
# 3.designation：葡萄酒酿造厂所在的葡萄园
# 4. points：分数
# 5. price：价格
# 6. province：葡萄酒来自的省份或州
# 7.region_1：区域1
# 8.region_2：区域2
# 9. variety：用来制作葡萄酒的葡萄品种
# 10. winery：酿酒的酿酒厂
df.info()
df.describe()
df.columns
```

```python
# 需求1：绘制图形，展示 产 葡萄酒最多的10个产地的信息.
# 3. 分组统计
# 思路1：根据产地(province)分组，基于points列(其他不包含空值或缺失值的列) 进行count 统计即可.
df.groupby('province', as_index=False).points.count().sort_values('points', ascending=False).head(10)

# 思路2： value_counts() 值的个数， 且会自动降序排列，它相当于：groupby() + count() + sort_values()
df['province'].value_counts().head(10)
```

```python
# 4. 针对于上述的结果，绘图展示
df['province'].value_counts().head(10).plot()  # 折线图 语法糖
df['province'].value_counts().head(10).plot(
    kind='line',
    figsize=(10, 5),
    color=['r', 'g', 'b', 'y', 'c', 'm', 'k', 'pink', 'gray']
) # 折线图
df['province'].value_counts().head(10).plot(
    kind='bar',
    figsize=(10, 5),
    color=['r', 'g', 'b', 'y', 'c', 'm', 'k', 'pink', 'gray']
) # 柱状图
```

```python
# 5. 最终写法
# step1:定义变量，记录参数，字典形式
text_kwargs = dict(figsize=(10, 5), color=['r', 'g', 'b', 'y', 'c', 'm', 'k', 'pink', 'gray'])
print(text_kwargs)

# step2: 绘图, 传参
df['province'].value_counts().head(10).plot(kind='bar', **text_kwargs)
```

```python
# 需求2：绘制图形，展示葡萄酒最多的10个产地的占比
# 1. 计算 葡萄酒最多的10个产地.
df['province'].value_counts().head(10)
# 2. 计算每个产地的占比, 即 产地产的葡萄酒数量/ 总共的葡萄酒数量
len(df)  # 150930
(df['province'].value_counts().head(10)) / len(df)

# 3. 绘图
(df['province'].value_counts().head(10) / len(df)).plot(kind='bar', **text_kwargs)
```

```python
# 需求3： 展示每个评分的普通酒种类(个数)， 即：80分 -->多少种酒， 90分 --->多少种酒 ....
df['points'].value_counts().sort_index().plot(**text_kwargs) # 折线图
df['points'].value_counts().sort_index().plot(kind='area',**text_kwargs) # 面积图
df['points'].value_counts().sort_index().plot(kind='bar',**text_kwargs) # 柱状图
```

## 2. 扩展_Seaborn绘图

```python
# 介绍 Seaborn 是一个 Python 数据可视化库，它基于 Matplotlib， 并且它提供了更高级的绘图功能.
# Pandas中绘图直接写单词，Seaborn中绘图是 图形名 + plot(), 例如: histplot()
```

```python
# 1. 加载数据
df = pd.read_csv('./data/tips.csv')  # 离线方式
# df = sns.load_dataset('tips')  # 在线方式(电脑要联网)， tips是 Seaborn 自带的数据集
# total_bill：总消费金额,tip：消费,sex：性别,smoker：是否抽烟,day：星期,time：就餐类型(午餐，晚餐),size：就餐人数
df
```

- 绘制图形，展示男女总账单的情况

```python
# 需求：绘制图形，展示男女总账单的情况
# 1. 创建画布
fig, ax = plt.subplots(figsize=(10, 5))
# 2. 绘制图形
# 参1：data 数据集  参2：x x轴列名  参3：hue 分组依据的列名
sns.histplot(data=df, x='total_bill', hue='sex')
# 3. 添加标题
ax.set_title('男女总账单的情况', fontsize=15)
# 4. 展示
plt.show()
```

- 散点图 scatter(), regplot(), jointplot()

```python
# 需求：描述 总账单和消费的分布情况
# 1. 创建画布
fig, ax = plt.subplots(figsize=(10, 5))
# 2. 绘制图形
# 思路1：散点图， scatter()
# 参1：data 数据集  参2：x x轴列名 总账单  参3：y Y轴的列名 消费  参4：hue 分组依据的列名
# sns.scatterplot(data=df, x='total_bill', y='tip', hue='sex')

# 思路2： regplot() 可以绘制散点图， 且可以绘制 拟合回归线(底层是：线性回归)
# sns.regplot(data=df, x='total_bill', y='tip')
# sns.regplot(data=df, x='total_bill', y='tip', fig_reg=True) # 绘制回归线
# sns.regplot(data=df, x='total_bill', y='tip', fit_reg=False) # 不绘制拟合回归线

# 思路3： joinplot() 可以绘制散点图， 且可以绘制 拟合回归线(底层是：线性回归)
sns.jointplot(data=df, x='total_bill', y='tip', kind='hex') # kind='hex' --> 蜂巢图

# 3. 添加标题
ax.set_title('总账单 和 小费的分布情况', fontsize=15)
# 4. 展示
plt.show()
```

- 箱线图，boxplot(), violinplot()

```python
# 绘制箱线图(也叫：盒须图)，展示：时间(time) 和账单总金额(total_bill)的分布情况
sns.boxplot(data=df, x='time', y='total_bill')
plt.show()
# IQR:四分位距 Inter Quartile Range
# IQR：Q3(75%)- Q1(25%)
```

```python
# 需求：查看 性别(sex) 和账单总金额(total_bill)，时间(time) 的分布情况
# 小提琴图, 适用于：箱线图 + 密度图
sns.violinplot(data=df, x='time', y='total_bill', hue='sex', split=True)
plt.show()
```
